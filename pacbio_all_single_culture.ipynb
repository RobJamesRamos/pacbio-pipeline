```r
wdpath <- getwd()
invam <-  file.path(wdpath, "ku-invam/kansas_rep_seqs_single_culture.fasta")
zuzana <- file.path(wdpath, "zuzana/zuzana_rep_seqs_single_culture.fasta")
zuzanars2 <- file.path(wdpath, "zuzana/zuzana_RSII_rep_seqs_single_culture.fasta")
system2("cat", args = c(invam, zuzana, zuzanars2), stdout = "combined/all_rep_seqs_single_culture.fasta")
```

if same species and culture, keep 1; if same species different culture, keep both; if different species, remove

```r
# Read combined file in
all_reads <- bs$readDNAStringSet("combined/all_rep_seqs_single_culture.fasta")

# Read metadata for combined files
metadata <- read.table("combined/all_asv_metadata.tsv", sep = "\t", header = TRUE)

# Find duplicates, this R function only returns one name for each duplicate pair
duplicates <- all_reads[bs$duplicated(all_reads), ]

# Match each duplicate to it's pair
duplicate_pairs <- data.frame()
for(x in names(duplicates)) {
  duplicate_pairs <- rbind(
    cbind(
      data.frame("ASV1" = names(duplicates[x])),
      data.frame(
        "ASV2" = (all_reads[duplicates[x] |> bs$match(all_reads)] |> names())
      )
    ),
    duplicate_pairs
  )
}

# Add in species and culture names from the metadata file
duplicate_pairs <- merge(
  duplicate_pairs,
  dplyr::rename_with(
    metadata[c("name", "species", "cultures")],
    ~ paste(., "1", sep = "_")
  ),
  by.x = "ASV1",
  by.y = "name_1"
)
duplicate_pairs <- merge(
  duplicate_pairs,
  dplyr::rename_with(
    metadata[c("name", "species", "cultures")],
    ~ paste(., "2", sep = "_")
  ),
  by.x = "ASV2",
  by.y = "name_2"
)
duplicate_pairs |> write.table(
  "combined/all_rep_seqs_single_culture_duplicate.tsv",
  sep = "\t"
)

# Write out all the duplicates into a fasta for record keeping
all_reads[
  names(all_reads) %in% as.vector(as.matrix(
    duplicate_pairs[c("ASV1", "ASV2")]
  ))
] |> bs$writeXStringSet("combined/all_rep_seqs_single_culture_duplicates.fasta")

# Write out the fasta of the sequences that had no duplicates
all_reads[
  !(names(all_reads) %in% as.vector(as.matrix(
    duplicate_pairs[c("ASV1", "ASV2")]
  )))
] |> bs$writeXStringSet("combined/all_rep_seqs_single_culture_deduplicates.fasta")

# Write out all duplicates that do not share a species, these are the true
# duplicates that should be dropped.
all_reads[
  names(all_reads) %in% as.vector(as.matrix(filter(
    duplicate_pairs,
    species_1 != species_2
  )[c("ASV1", "ASV2")]))
] |> bs$writeXStringSet("combined/all_rep_seqs_single_culture_true_duplicates.fasta")

# Write out one of each pair of duplicates that share both species and culture
all_reads[
  names(all_reads) %in% filter(
    duplicate_pairs,
    species_1 == species_2
    & cultures_1 == cultures_2
  )[c("ASV2")]
] |> bs$writeXStringSet("combined/all_rep_seqs_single_culture_same_duplicates.fasta")

# Write out both sequences for duplicates that share species, but not culture
all_reads[
  names(all_reads) %in% as.vector(as.matrix(filter(
    duplicate_pairs,
    species_1 == species_2
    & cultures_1 != cultures_2
  )[c("ASV1", "ASV2")]))
] |> bs$writeXStringSet("combined/all_rep_seqs_single_culture_diff_duplicates.fasta")

# Read each file back in
dedup <- bs$readDNAStringSet(
  "combined/all_rep_seqs_single_culture_deduplicates.fasta"
)
same <- bs$readDNAStringSet(
  "combined/all_rep_seqs_single_culture_same_duplicates.fasta"
)
diff <- bs$readDNAStringSet(
  "combined/all_rep_seqs_single_culture_diff_duplicates.fasta"
)

# combine the deduplicated files with the false duplicates
# i.e same species/same culture, and same species/different culture.
bs$writeXStringSet(
  c(dedup, same, diff),
  "combined/all_rep_seqs_single_culture_unique.fasta"
)
```

There could have been an issue if we had a set of duplicates with one pair with the same species and one pair without, but since we don't I think we're ok.
